```python
import pandas as pd
```

```python
df = pd.read_csv("Dataset1.csv")
```

```python
print("Total Records:", df.shape[0])
```

```python
print("Total Attributes:", df.shape[1])
```

```python
print("\nColumn Names:")
print(df.columns.tolist())
```

```python
print("\nDataset Info:")
print(df.info())
```

```python
train_routes = (
    df.sort_values(['Train_No', 'SN'])
      .groupby('Train_No')
      .agg(
          Starting_Station=('Station_Name', 'first'),
          Ending_Station=('Station_Name', 'last')
      )
      .reset_index()
)

print(train_routes.head())
```

```python
stops_per_train = (
    df.groupby('Train_No')
      .size()
      .reset_index(name='Number_of_Stops')
)

print(stops_per_train.head())
```

```python
max_stops = stops_per_train.loc[
    stops_per_train['Number_of_Stops'].idxmax()
]

min_stops = stops_per_train.loc[
    stops_per_train['Number_of_Stops'].idxmin()
]

print("Train with Maximum Stops:")
print(max_stops)

print("\nTrain with Minimum Stops:")
print(min_stops)
```

```python
print("===== DATASET SUMMARY =====")
print("Records:", df.shape[0])
print("Attributes:", df.shape[1])

print("\nTotal Trains:", df['Train_No'].nunique())

print("\nTrain with Maximum Stops:")
print(max_stops)

print("\nTrain with Minimum Stops:")
print(min_stops)
```
